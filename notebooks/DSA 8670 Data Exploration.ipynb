```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("AmesHousing-1.csv")
df.head()
df.describe()
```

```python
plt.hist(df["Sale Type"], bins = 10, edgecolor="black")
plt.title("Distribution of Home Sale Types")
plt.xlabel("Sale Type")
plt.ylabel("Frequency")
```

```python
plt.scatter(df["Lot Area"], df["SalePrice"])
plt.title("Lot Area vs. Sale Price")
plt.xlabel("Lot Area")
plt.ylabel("Sale Price ($)")
```
