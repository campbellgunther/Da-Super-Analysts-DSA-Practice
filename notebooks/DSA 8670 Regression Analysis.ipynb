```python
import pandas as pd
```

```python
df = pd.read_csv("../data/AmesHousing-1.csv")
df.head()
```

```python
import statsmodels.api as sm

X = df[["Lot Area", "MS SubClass", "Yr Sold"]]
y = df["SalePrice"]

X = sm.add_constant(X)

model = sm.OLS(y, X).fit()
print(model.summary())
```
